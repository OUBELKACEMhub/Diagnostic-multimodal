# 02 - EDA et visualisation

Inspection des fichiers valides sans charger l'ensemble des pixels en memoire : distribution des classes et un exemple par classe.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data_preprocessing import inspecter_dataset_images

images, fichiers_invalides = inspecter_dataset_images(PROJECT_ROOT / "data" / "raw")
print(f"{len(images)} images valides ; {len(fichiers_invalides)} fichier(s) invalide(s) ignore(s).")
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=images, x="label", order=sorted(images["label"].unique()))
plt.title("Repartition des images valides par classe")
plt.xlabel("Classe")
plt.ylabel("Nombre d'images")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
classes = list(images.groupby("label", sort=True))
fig, axes = plt.subplots(1, len(classes), figsize=(4 * len(classes), 4), squeeze=False)
for ax, (classe, groupe) in zip(axes.flat, classes):
    with Image.open(groupe.iloc[0]["path"]) as image:
        ax.imshow(image.convert("RGB"))
    ax.set_title(classe)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
